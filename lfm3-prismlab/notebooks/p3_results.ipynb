```python

# Manual labels from the six QA results
tests = [
    ("answerable", True),   # T1
    ("answerable", False),  # T2
    ("answerable", True),   # T3
    ("answerable", True),   # T4
    ("unanswerable", True), # T5
    ("unanswerable", False) # T6
]

for category in ("answerable", "unanswerable"):
    scores = [passed for kind, passed in tests
              if kind == category]

    accuracy = sum(scores) / len(scores)
    print(f"{category}: {accuracy:.1%}")
    print(scores)
print(sum(scores))



```
