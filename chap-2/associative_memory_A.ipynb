```python
# Start with fresh memory
S = torch.zeros(2, 3)

k = torch.tensor([1.0, 0.0])
v = torch.tensor([2.0, -1.0, 3.0])

beta = 0.5

prediction = k @ S

# Prediction error — no beta here
error = v - prediction

# Delta-rule update
S = S + beta * torch.outer(k, error)

retrieved = k @ S

print("Updated memory:\n", S)
print("Retrieved:", retrieved)
```

```python
import torch

S = torch.zeros(2, 3)

k = torch.tensor([1.0, 0.0])
v = torch.tensor([2.0, -1.0, 3.0])
beta = 0.5

for step in range(2):
    prediction = k @ S
    error = v - prediction

    S = S + beta * torch.outer(k, error)

    print(f"Update {step + 1}: {k @ S}")
```

```python
S = torch.zeros(2, 3)

k = torch.tensor([1.0, 0.0])

v1 = torch.tensor([2.0, -1.0, 3.0])
v2 = torch.tensor([5.0, 4.0, -2.0])

beta = 1.0

def update_memory(S, k, v, beta):
    prediction = k @ S
    error = v - prediction
    return S + beta * torch.outer(k, error)

# Store the first association
S = update_memory(S, k, v1, beta)
print("After V1:", k @ S)

# Store a different value under the SAME key
S = update_memory(S, k, v2, beta)
print("After V2:", k @ S)

print("V1 preserved?", torch.allclose(k @ S, v1))
print("V2 retrieved?", torch.allclose(k @ S, v2))
```
